## 2. The question: decision, action, cost of a wrong call

The decision is: **which pages should a content or SEO reviewer look at first for a possible CTR or engagement opportunity?**

The output will be a ranked list of pages with an opportunity score and reason codes. A reviewer could then inspect the highest-ranked pages and decide whether to improve the title or metadata, improve intent matching or snippet structure, investigate engagement, or simply monitor the page.

A wrong recommendation has two costs. If I rank a weak opportunity too highly, the reviewer may waste limited editing time on a page that does not need attention. If I rank a genuine opportunity too low, a page that could potentially capture more clicks may remain unreviewed. Therefore, the system should prioritize evidence and minimum-volume requirements rather than simply ranking every low-CTR page.

## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
import pandas as pd

df = pd.read_csv("data/raw/content_refresh_anonymized.csv")

print("Starter dataset rows:", len(df))

visible = df[df["impressions_90d"] >= 100]

ctr_by_position = (
    visible.groupby("position_tier")["ctr"]
    .mean()
    .sort_values(ascending=False)
)

print("\nMean CTR by position tier (impressions >= 100):")
print(ctr_by_position.round(4))

correlation = df[["avg_position", "ctr"]].corr().loc["avg_position", "ctr"]
print(f"\nRaw avg_position vs CTR correlation: {correlation:.3f}")
```

## 4. Careful words: what I can and can't claim

I can claim that the starter data shows an observed directional relationship between CTR and position tiers, and that some pages may have lower CTR than comparable pages after accounting for position. I can use these signals to build a decision-support ranking for pages that deserve human review.

I cannot claim that position causes the observed CTR differences, that a particular title or metadata change will definitely increase clicks, or that the analysis has discovered Google's ranking algorithm. The results are observational. Any future scoring or model will need minimum-volume filters, leakage checks, and validation before being treated as a reliable review tool.


```python
# Check the evidence used in my framing
assert len(df) == 30000
assert "ctr" in df.columns
assert "position_tier" in df.columns
assert "avg_position" in df.columns

print("Evidence check passed.")
print(f"Rows analysed: {len(df)}")
print(f"Raw avg_position vs CTR correlation: {correlation:.3f}")
```
